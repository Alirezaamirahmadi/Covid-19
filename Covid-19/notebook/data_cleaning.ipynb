```python
# وارد کردن کتابخانه 
import pandas as pd

# خواندن دیتاست
df = pd.read_csv("full_grouped.csv")

# تبدیل ستون از متن به تاریخ
df["Date"] = pd.to_datetime(df["Date"])
```

```python
# بررسی مقادیر گمشده 

print("Missing Values:")
print(df.isnull().sum())

# اگر مقدار گمشده وجود داشته باشد، تعداد آن‌ها را نمایش می‌دهیم
missing_values = df.isnull().sum().sum()

print("Number of Missing Values:", missing_values)
```

```python
# رکوردهای تکراری

rows_count = df.duplicated().sum()

print("Duplicates:", rows_count)

# حذف رکوردهای تکراری
df = df.drop_duplicates()

print("After Remove Duplicates:", len(df))

```

```python
# بررسی تکراری بودن کشور و تاریخ

country_date = df.duplicated(subset=["Country/Region", "Date"]).sum()

print("Duplicates:", country_date)
```

```python
# بررسی و حذف مقادیر منفی

numeric_cols = [
    "Confirmed", "Deaths", "Recovered", "Active",
    "New cases", "New deaths", "New recovered"
]

print("Negative Before Remove:")

for col in numeric_cols:
    negative_count = (df[col] < 0).sum()
    print(col, ":", negative_count)


# حذف رکوردهایی که در ستون‌های عددی مقدار منفی دارند
negative_rows = (df[numeric_cols] < 0).any(axis=1)

print("Negative Rows:",
      negative_rows.sum())

df = df[~negative_rows].copy()

print("Number of Negative Rows:",
      len(df))


# بررسی دوباره
print("Negative After Remove:")

for col in numeric_cols:
    negative_count = (df[col] < 0).sum()
    print(col, ":", negative_count)
```

```python
# بررسی سازگاری ستون اکتیو

# Confirmed - Deaths - Recovered
calculated_active = (
    df["Confirmed"] - df["Deaths"] - df["Recovered"])

active_errors = (df["Active"] != calculated_active).sum()

print("Incompatible:", active_errors)
```

```python
#بررسی نهایی داده

print("Final Dataset:", df.shape)

print("Missing Value after Cleaning:")
print(df.isnull().sum())

print("Duplicate Value after Cleaning:",
      df.duplicated().sum())
```

```python
# ذخیره دیتاست

df.to_csv("cleaned_covid_data.csv", index=False)
```
